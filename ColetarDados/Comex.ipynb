```python
import pandas as pd
from pathlib import Path
```

```python
caminho_arquivo = Path("..") / "DadosBrutos" / "V_EXPORTACAO.csv"
```

```python
df = pd.read_csv(caminho_arquivo, sep=';', encoding='utf-8-sig')

colunas_texto = ['Países', 'Descrição NCM', 'UF do Produto']
for col in colunas_texto:
    df[col] = df[col].str.replace(r'[\r\n]+', ' ', regex=True).str.strip()
```

```python
colunas = ['Ano', 'UF do Produto', 'Código NCM', 'Descrição NCM', 'Valor US$ FOB', 'Países']
angola = df.loc[df['Países'] == 'Angola', colunas]

pd.set_option('display.max_rows', None)   
angola.sort_values('Valor US$ FOB', ascending=False)
```

```python
sp = df[(df['Países'] == 'Angola') &
        (df['UF do Produto'] == 'São Paulo') &
        (df['Ano'] == 2014)]

sp[['Ano', 'Países', 'Código NCM', 'Descrição NCM', 'UF do Produto', 'Valor US$ FOB']] \
    .sort_values(['Código NCM', 'Valor US$ FOB'], ascending=[True, False])
```
