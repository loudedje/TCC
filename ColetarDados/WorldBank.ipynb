```python
import wbgapi as wb
import pandas as pd
```

```python
# Países PALOP e Brasil
paises = ['AGO', 'MOZ', 'CPV', 'GNB', 'STP', 'BRA']

# Indicadores
indicadores = {
    'SP.POP.TOTL':       'População total',
    'NY.GDP.PCAP.CD':    'PIB per capita (USD)',
    'NY.GDP.PCAP.PP.CD': 'PIB per capita PPP (USD)',
    'SE.ADT.LITR.ZS':    'Taxa de alfabetização (%)',
    'NY.GDP.MKTP.KD.ZG': 'Crescimento PIB (%)'
}

dfs = []
```

```python
for cod, nome in indicadores.items():
    df = wb.data.DataFrame(cod, paises, time=2025)
    df = df.reset_index()
    df.columns = ['iso_code', 'valor']
    df['ano'] = 2025
    df['indicador'] = nome
    dfs.append(df)

df_final = pd.concat(dfs, ignore_index=True)
df_final = df_final.dropna(subset=['valor'])

tabela = df_final.pivot_table(
    index='indicador',
    columns='iso_code',
    values='valor',
    aggfunc='first'
)

print(tabela.round(2).to_string())
```
