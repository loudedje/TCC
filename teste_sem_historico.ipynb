```python
import pandas as pd
from pathlib import Path
import pandas as pd
```

```python
catalogo = pd.read_excel("catalogo_.xlsx")
catalogo.head()
```

```python
caminho_arquivo = Path(".") / "DadosBrutos" / "V_EXPORTACAO.csv"
df = pd.read_csv(caminho_arquivo, sep=';', encoding='utf-8-sig')

colunas_texto = ['Países', 'Descrição NCM', 'UF do Produto']
for col in colunas_texto:
    df[col] = df[col].str.replace(r'[\r\n]+', ' ', regex=True).str.strip()

df['Código NCM'] = df['Código NCM'].astype(str).str.zfill(8)
```

```python
catalogo['NCM_normalizado'] = catalogo['NCM'].astype(str).str.replace('.', '', regex=False)
catalogo.loc[catalogo['NCM_normalizado'] == '21069029', 'NCM_normalizado'] = '21069029'

print(catalogo['NCM_normalizado'].str.len().value_counts())   
print(catalogo['NCM_normalizado'].nunique())                   
```

```python
ncms_catalogo = catalogo['NCM_normalizado'].unique()
ncms_com_historico = df['Código NCM'].unique()

sem_historico = []
com_historico = []

for ncm in ncms_catalogo:
    if ncm in ncms_com_historico:
        com_historico.append(ncm)
    else:
        sem_historico.append(ncm)

print("Total no catálogo:", len(ncms_catalogo))
print("Com histórico:", len(com_historico))
print("Sem histórico:", len(sem_historico))
print(sem_historico)
```
